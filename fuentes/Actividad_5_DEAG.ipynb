```python
import os
import time
import subprocess
from pyngrok import ngrok

# 1. Limpiar todo rastro de intentos anteriores
os.system("fuser -k 5000/tcp")
ngrok.kill()

print("Iniciando MLflow y conectando base de datos...")
# 2. Iniciar MLflow guardando el registro de errores
proceso_mlflow = subprocess.Popen(
    "mlflow ui --backend-store-uri sqlite:///mlflow.db --host 127.0.0.1 --port 5000",
    shell=True,
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)

# 3. Le damos 6 segundos reales para que la base de datos cargue
time.sleep(6)

# 4. Conectamos Ngrok
try:
    public_url = ngrok.connect(addr="127.0.0.1:5000", host_header="127.0.0.1:5000")
    print(f"👉 Listo. Clic aquí: {public_url}")
except Exception as e:
    print("Fallo al crear el túnel.")

# Verificamos si MLflow se apagó por algún error interno
if proceso_mlflow.poll() is not None:
    _, errores = proceso_mlflow.communicate()
    print("\n⚠️ ERROR INTERNO DE MLFLOW:")
    print(errores.decode('utf-8'))
```

```python
import os
import time
from pyngrok import ngrok

# 1. Apagar el escudo de seguridad estricto de MLflow
os.environ["MLFLOW_SERVER_DISABLE_SECURITY_MIDDLEWARE"] = "true"

# 2. Limpiar todos los procesos colgados
os.system("pkill -9 -f mlflow")
os.system("fuser -k 5050/tcp")
ngrok.kill()

print("Iniciando MLflow con seguridad de red desactivada...")

# 3. Levantar el servidor
ruta_db = "sqlite:////content/Actividad_5_DEAG/mlflow.db"
get_ipython().system_raw(f"mlflow ui --backend-store-uri {ruta_db} --host 0.0.0.0 --port 5050 &")

time.sleep(4)

# 4. Generar el túnel final
try:
    public_url = ngrok.connect(addr="5050")
    print(f"👉 ENLACE DEFINITIVO: {public_url}")
except Exception as e:
    print(f"Error al conectar Ngrok: {e}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ajusta el nombre al archivo real presente en la carpeta
df_eda = pd.read_csv("datos/datos_ini/WA_Fn-UseC_-Telco-Customer-Churn.csv")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gráfica 1: Abandono por Contrato
sns.countplot(data=df_eda, x='Contract', hue='Churn', ax=axes[0], palette='viridis')
axes[0].set_title('Tasa de Abandono por Tipo de Contrato')

# Gráfica 2: Cargos Mensuales vs Abandono
sns.boxplot(data=df_eda, x='Churn', y='MonthlyCharges', ax=axes[1], palette='viridis')
axes[1].set_title('Distribución de Cargos Mensuales por Abandono')

plt.tight_layout()
plt.show()
```
